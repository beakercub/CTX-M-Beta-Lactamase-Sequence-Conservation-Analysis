```python
from pathlib import Path
import subprocess
```

```python
data_dir = Path("../data")

input_fasta = data_dir / "ctxm_refseq_full.fasta"
output_alignment = data_dir / "ctxm_refseq_full_aligned.fasta"

input_fasta, output_alignment
```

```python
cmd = [
    "clustalo",
    "-i", str(input_fasta),
    "-o", str(output_alignment),
    "--force",
    "--verbose"
]

import subprocess
subprocess.run(cmd)
```
