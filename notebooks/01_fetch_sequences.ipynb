```python
from Bio import Entrez, SeqIO
from pathlib import Path
```

```python
Entrez.email = "wangy994@mcmaster.com"
```

```python
query = "CTX-M beta-lactamase AND refseq[filter] AND 250:300[slen]"
query
```

```python
handle = Entrez.esearch(
    db="protein",
    term=query,
    retmax=200
)

record = Entrez.read(handle)
protein_ids = record["IdList"]

len(protein_ids)
```

```python
record["Count"]
```

```python
#direct to data file
data_dir = Path("../data")
```

```python
handle = Entrez.efetch( #access NCBI Entrez
    db="protein", #protein library
    id=protein_ids,
    rettype="fasta", # return fasta file
    retmode="text"
)

sequences = list(SeqIO.parse(handle, "fasta"))
len(sequences) # check the protein sequence number
```

```python
# write sequence into fasta
out_fasta = data_dir / "ctxm_refseq_full.fasta"

SeqIO.write(sequences, out_fasta, "fasta")

out_fasta
```
