```python
from pathlib import Path
from Bio import AlignIO
import numpy as np
import matplotlib.pyplot as plt
```

```python
## read the generated aligned fasta file 

data_dir = Path("../data")
aligned_fasta = data_dir / "ctxm_refseq_full_aligned.fasta"

aligned_fasta.exists(), aligned_fasta
```

```python
alignment = AlignIO.read(aligned_fasta, "fasta")
type(alignment)
print("number of protein sequences are:",len(alignment), ", and the number of positions after alignment for each sequence is:", alignment.get_alignment_length())
```

```python
alignment_array = np.array([list(rec.seq) for rec in alignment])
alignment_array.shape
```

```python
## find out the most frequet amino acid in each alignment spot, and take the counts
def conservation_score(column):
    residues, counts = np.unique(column, return_counts=True)
    return counts.max() / counts.sum()
```

```python
conservation = np.array([
    conservation_score(alignment_array[:, i])
    for i in range(alignment_array.shape[1])
])
```

```python
##plot
plt.figure(figsize=(12, 4))
plt.plot(conservation)
plt.xlabel("Alignment position")
plt.ylabel("Conservation score")
plt.title("Conservation profile of CTX-M beta-lactamase")
plt.tight_layout()
plt.show()
```
